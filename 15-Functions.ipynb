```python
# args / kwargs
```

```python
def args_sum(*args):
    print(sum(args))
```

```python
args_sum(10,30,50)
```

```python
def args_sum(*args):
    y = 0
    for x in args:
        y += x
    return y
```

```python
args_sum(10,20,30)
```

```python
def kwargs_excample(**kwargs):
    print(kwargs)
```

```python
kwargs_excample(banana=200, orange=400, melon=600)
```

```python
try:
    myAge = int(input("enter age: "))
    print(myAge * 2)
except:
    print("enter your age!!!")
```

```python
while True:
    try:
        myAge = int(input("enter age: "))
        print(myAge * 2)
        break
    except:
        print("enter your age!!!")
```

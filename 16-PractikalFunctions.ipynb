```python
def divideNumber(number):
    return number/2
```

```python
myList = 3,5,7,10,20,30
```

```python
newList = []
for number in myList:
    newList.append(divideNumber(number)) 
print(newList)
```

### map

```python
list(map(divideNumber, myList))
```

### scope

#### LEGB -> L->Local, E->Enclosing, G->Global, B->Built-in

```python
#Global
myString = "oguz"

def myFunction():
    #Enclosing
    myString = "oguz 2"
    print(myString)
    
    def myFunction2():
        #Local
        myString = "oguz 3"
        print(myString)
    
    myFunction2()
```

```python
myString
```

```python
myFunction()
```

```python
y = 10
def chancY():
    global y
    y = 5
    return y * 5
```

```python
chancY()
```

# inheritance

```python
class Musician():
    
    def __init__(self, name):
        self.name = name
        print('musician class')

    def test1(self):
        print('test1')

    def test2(self):
        print('test2')
```

```python
oguz = Musician('oguz')
```

```python
oguz.name
```

```python
oguz.test1()
```

```python
class MusicianPlus(Musician):

    def __init__(self,name):
        Musician.__init__(self,name)
        print('musician plus')

    def test3(self):
        print('test3')

    def test1(self):
        print('test1 modified')
```

```python
batuhan = MusicianPlus('batuhan')
```

```python
batuhan.test1()
```

```python
batuhan.test2()
```

```python
batuhan.name ='oguz batuhan'
```

```python
type(batuhan)
```

```python
batuhan.test3()
```

```python
batuhan.name
```

```python
batuhan.test1()
```

```python
oguz.test1()
```

# polymorphism

```python
class Banana():
    def __init__(self,name):
        self.name = name
    def info(self):
        return f'100 kcal {self.name}'

class Apple():
    def __init__(self,name):
        self.name = name
    def info(self):
        return f'150 kcal {self.name}'
```

```python
banana = Banana('banana')
```

```python
apple = Apple('apple')
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruitList = [banana, apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

# encapsulation

```python
class Phone():
    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def info(self):
        print(f'{self.name} price is: {self.__price}')

    def changePrice(self,price):
        self.__price = price
```

```python
iphone = Phone('iphone 18', 3000)
```

```python
iphone.info()
```

```python
iphone.__price = 2500
```

```python
iphone.info()
```

```python
iphone.changePrice(4000)
```

```python
iphone.info()
```

# abstraction

```python
 from abc import ABC, abstractmethod
```

```python
class Car(ABC):
    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
class Tesla(Car):

    def maxSpeed(self):
        print('200km')
```

```python
tesla = Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):

    def maxSpeed(self):
        print('250km')
```

```python
mercedes = Mercedes()
```

```python
mercedes.maxSpeed()
```

```python
class Fruit():

    def __init__(self,name,calories):
        self.name = name
        self.calories = calories

    def __str__(self):
        return f'{self.name}: {self.calories} calories'

    def __len__(self):
        return self.calories
```

```python
myFruit = Fruit('banana', 150)
```

```python
myFruit.name
```

```python
print(myFruit)
```

```python
len(myFruit)
```
